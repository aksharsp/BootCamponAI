```python
import random
random_numbers = [random.randint(1, 100) for _ in range(10)]

print("Original List of Random Numbers:", random_numbers)

print("Even numbers in the list:")
for num in random_numbers:
  if num % 2 == 0:
    print(num)
```

```python
original_string = "My Name is Akshar"
print("Original String:", original_string)

reversed_characters = original_string[::-1]
print("Reversed Characters:", reversed_characters)

words = original_string.split()
reversed_words = words[::-1]
output_words = " ".join(reversed_words)
print("Reversed Words:", output_words)
```
