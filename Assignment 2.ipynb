```python
import random
random_numbers = [random.randint(1, 100) for _ in range(10)]

print("Original List of Random Numbers:", random_numbers)

print("Even numbers in the list:")
for num in random_numbers:
  if num % 2 == 0:
    print(num)
```

```python
original_string = "My Name is Akshar"
print("Original String:", original_string)

reversed_characters = original_string[::-1]
print("Reversed Characters:", reversed_characters)

words = original_string.split()
reversed_words = words[::-1]
output_words = " ".join(reversed_words)
print("Reversed Words:", output_words)
```

```python
import math
import matplotlib.pyplot as plt

points = 100
x_values = [i * (2 * math.pi / points) for i in range(points + 1)]
y_values = [math.sin(x) for x in x_values]

fig, ax = plt.subplots(figsize=(8, 4))

ax.plot(x_values, y_values, color="darkviolet", linewidth=2.5, label="y = sin(x)")
ax.axhline(0, color='black', linewidth=1) # Adds a clean X-axis center line
ax.grid(True, linestyle="--", alpha=0.5)

ax.set_title("Trigonometric Sine Wave Graph", fontsize=14)
ax.set_xlabel("Angle (Radians)")
ax.set_ylabel("Amplitude")
ax.legend()

plt.tight_layout()
plt.show()
```
