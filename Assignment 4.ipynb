```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
image = np.zeros((400, 600, 3), dtype=np.uint8)
cv2.rectangle(image, (50, 50), (550, 350), (200, 200, 200), -1)
cv2.circle(image, (300, 200), (80), (0, 0, 255), -1)
cv2.putText(image, 'OpenCV Lab', (180, 220), cv2.FONT_HERSHEY_SIMPLEX, 1.5, (255, 255, 255), 3)

plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))
plt.title('Original Image')
plt.axis('off')
plt.show()
```

```python
gray_image = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

h, w, _ = image.shape
cropped_image = image[50:350, 100:500]

plt.figure(figsize=(10, 5))
plt.subplot(1, 2, 1)
plt.imshow(gray_image, cmap='gray')
plt.title('Grayscale Conversion')
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(cv2.cvtColor(cropped_image, cv2.COLOR_BGR2RGB))
plt.title('Cropped Region')
plt.axis('off')
plt.show()
```

```python
resized_image = cv2.resize(image, (300, 200))

center = (w // 2, h // 2)
rotation_matrix = cv2.getRotationMatrix2D(center, 45, 1.0)
rotated_image = cv2.warpAffine(image, rotation_matrix, (w, h))

plt.figure(figsize=(10, 5))
plt.subplot(1, 2, 1)
plt.imshow(cv2.cvtColor(resized_image, cv2.COLOR_BGR2RGB))
plt.title('Resized Down')

plt.subplot(1, 2, 2)
plt.imshow(cv2.cvtColor(rotated_image, cv2.COLOR_BGR2RGB))
plt.title('Rotated 45 Degrees')
plt.axis('off')
plt.show()
```

```python
blurred_image = cv2.GaussianBlur(image, (15, 15), 0)
edges = cv2.Canny(gray_image, 100, 200)

plt.figure(figsize=(10, 5))
plt.subplot(1, 2, 1)
plt.imshow(cv2.cvtColor(blurred_image, cv2.COLOR_BGR2RGB))
plt.title('Gaussian Blur')
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(edges, cmap='gray')
plt.title('Canny Edge Detection')
plt.axis('off')
plt.show()
```
